```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR", "Finance"],
    "Salary": [60000, 50000, 65000, 52000, 70000],
    "Gender": ["M", "F", "F", "M", "F"]
})

```

```python
df.info()
```

```python
df.head()
```

```python
df.groupby('Department')['Gender'].value_counts()
```

```python
df.groupby('Department')['Salary'].mean()
```

```python
df.groupby(['Department','Gender']).size().reset_index(name = 'Count')
```

```python
df.groupby('Department')['Salary'].mean().sort_values(ascending=False).head(n=3)
```

```python
df.groupby('Gender').agg(Salary =("Salary",lambda x:x.max()-x.min()))
```
