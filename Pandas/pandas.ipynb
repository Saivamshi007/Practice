```python
Swap every two adjacent number and return list. You must solve the problem without modifying the values in the list.
Example 1:
Input: list1 = [1,2,3,4]
Output: [2,1,4,3]
Example 2:
Input: list2 = [1,2,3]
Output: [2,1,3]
```

```python
nums = [1,2,3,4]
no_swaps = len(nums)//2
for i in range(0,no_swaps):
    j = 0
    nums[j],nums[j+1] = nums[j+1],nums[j]
    j+=2
print(nums)
```
